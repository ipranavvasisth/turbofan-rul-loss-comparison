```python
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error
from src.data import load_fd001, scale_features, last_rows
```

```python
train, test, test_rul, feats = load_fd001()
train, test, _ = scale_features(train, test, feats)

X_train, y_train = train[feats].values, train["RUL"].values
X_test = last_rows(test, feats)
y_test = np.minimum(test_rul, 125)   # clip test targets the same way as train

models = {
    "Ridge": Ridge(alpha=1.0),
    "RandomForest": RandomForestRegressor(n_estimators=200, random_state=42, n_jobs=-1),
}
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    print(name, "RMSE:", np.sqrt(mean_squared_error(y_test, pred)))
```
